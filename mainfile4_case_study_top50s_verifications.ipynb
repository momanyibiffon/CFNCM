```python
# this file is mainly used for case study of the top 50 breast and lung neoplasms
# using two authoritative data sets, dbDEMC and miR2Disease
import pandas as pd
import numpy as np
```

```python
# fetching the breast and lung neoplasms data from known associations with their association scores
eval_df = pd.read_csv('datasets/predicted_associations/evaluation_breast_lung_knownpred_scores_jan9.csv')
```

```python
eval_df
```

```python
# categorizing breast and lung neoplasms data
breast_df = eval_df[eval_df['Disease']==50]
lung_df = eval_df[eval_df['Disease']==236]
breast_df = breast_df.sort_values(by='Score', ascending=False)
lung_df = lung_df.sort_values(by='Score', ascending=False)
```

```python
breast_df, lung_df
```

```python
# getting top 50 samples
breast_df_top50 = breast_df.head(50)
lung_df_top50 = lung_df.head(50)
```

```python
breast_df_top50.shape, lung_df_top50.shape
```

```python
breast_df_top50
```

```python
lung_df_top50
```

```python
# reading miRNA names
mirna_names = pd.read_csv('datasets/1.miRNA_disease_association/miRNA name.csv', header=None)
```

```python
mirna_names
```

```python
breast_df_top50['mirna_name'] = ''
lung_df_top50['mirna_name'] = ''
```

```python
# fetching mirna names
for m in range(len(mirna_names)):
    for n in range(len(lung_df_top50)):
            if breast_df_top50.iloc[n, 0] == mirna_names.iloc[m, 0]:
                breast_df_top50.iloc[n, 3] = mirna_names.iloc[m, 1]
            if lung_df_top50.iloc[n, 0] == mirna_names.iloc[m, 0]:
                lung_df_top50.iloc[n, 3] = mirna_names.iloc[m, 1]
```

```python
# saving breast and lung neoplasms data with actual associated miRNA names
breast_df_top50.to_csv("datasets/predicted_associations/case_study/breast_df_top50_with_names.csv", index=False)
lung_df_top50.to_csv("datasets/predicted_associations/case_study/lung_df_top50_with_names.csv", index=False)
```
